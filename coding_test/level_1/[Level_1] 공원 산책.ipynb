```python
def solution(park, routes):
    for index, row in enumerate(park):
        if 'S' in row:
            x,y = row.index('S'), index
            break
    for nav in routes:
        direction, step = nav.split()
        if direction == "E":
            ny = y + int(step)
            if ny < len(park[0]) and 'X' not in park[x][y+1:ny+1]:
                y = ny
        if direction == 'W':
            ny = y - int(step)
            if ny >= 0 and 'X' not in park[x][ny:y]:
                x = ny
        if direction == "S":
            nx = x - int(step)
            if nx < len(park) and all(park[row][y] != 'X' for row in range(x+1,nx+1)):
                x = nx
        if direction == 'N':
            nx = x + int(step)
            if nx >= 0 and all(park[row][y] != 'X' for row in range(nx,x)):
                x = nx
    return y,x
            
# OOO
# OXX
# OOO


# solution(["SOO","OOO","OOO"], ["E 2","S 2","W 1"]) # [2,1]
solution(["SOO","OXX","OOO"], ["E 2","S 2","W 1"]) # [0,1]
# solution(["OSO","OOO","OXO","OOO"], ["E 2","S 3","W 1"]) # [0,0]
```
